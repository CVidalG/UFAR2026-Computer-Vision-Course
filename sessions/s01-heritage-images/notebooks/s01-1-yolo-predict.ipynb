```python
# --- 7. Inference parameters ---
# conf: minimum confidence score (default 0.25)
for conf in [0.5, 0.25, 0.05]:
    r = model.predict(source, conf=conf, verbose=False)[0]
    print(f"conf={conf}: {len(r.boxes)} box(es)")
```

```python
# iou: threshold of the non-maximum suppression (default 0.5): boxes of the same class overlapping more are merged
for iou in [0.2, 0.5, 0.9]:
    r = model.predict(source, conf=0.05, iou=iou, verbose=False)[0]
    print(f"iou={iou}: {len(r.boxes)} box(es)")
```

```python
# imgsz: size of the network input (longest side, default 640)
for imgsz in [320, 640, 1280]:
    r = model.predict(source, imgsz=imgsz, verbose=False)[0]
    print(f"imgsz={imgsz}: {len(r.boxes)} box(es), "
          f"inference {r.speed['inference']:.0f} ms")
```

```python
# Centres (x, y) of the boxes
centres = r.boxes.xywh[:, :2]
```

```python
centres[:5]
```

```python
# Median height of a character, in pixels
h = float(r.boxes.xywh[:, 3].median())
```

```python
h
```

```python
# --- Reading order, as on a page: (1) cut the characters into lines, from their height y;
# (2) read each line from left to right, by x; (3) take the lines from top to bottom.
# Assumed: roughly horizontal lines, a script read from left to right (Armenian).
# (1) The boxes sorted by increasing y, from the top of the image to the bottom
by_y = centres[:, 1].argsort().tolist()
```

```python
by_y[:10]
```

```python
# Their y, in this order: increasing
centres[by_y[:10], 1]
```

```python
# Jump of y from one box to the next in this order: small within a line, large between two lines
jumps = centres[by_y[1:], 1] - centres[by_y[:-1], 1]
```

```python
# Rounded to 0.1 pixel
jumps[:30].round(decimals=1)
```

```python
# A jump larger than half a character height (h / 2) starts a new line: number of such jumps
int((jumps > h / 2).sum())
```

```python
# Text lines, as lists of box indices
lines = []
```

```python
# The boxes taken by increasing y; a new line starts when y jumps by more than h / 2
for i in by_y:
    if not lines or centres[i, 1] - centres[lines[-1][-1], 1] > h / 2:
        lines.append([])
    lines[-1].append(i)
```

```python
len(lines)
```

```python
# Characters per line
[len(line) for line in lines]
```

```python
# (2) Within a line: the first line as found, ordered by y
lines[0][:8]
```

```python
# Their x, in this order: not from left to right
centres[lines[0][:8], 0]
```

```python
# The first line sorted by increasing x
first_line = sorted(lines[0], key=lambda i: float(centres[i, 0]))
```

```python
centres[first_line[:8], 0]
```

```python
# (3) Reading order: each line sorted by x, then the lines one after the other, from top to bottom
reading_order = [i for line in lines for i in sorted(line, key=lambda i: float(centres[i, 0]))]
```

```python
reading_order[:10]
```

```python
# Every box once: as many indices as boxes
len(reading_order)
```

```python
# Rank of each box in the reading order
rank = {i: k for k, i in enumerate(reading_order)}
```

```python
# Check on the image: the rank of each character, one colour per line
plt.figure(figsize=(12, 6.7))
plt.imshow(r.orig_img[:, :, ::-1])
for n, line in enumerate(lines):
    for i in line:
        x, y = centres[i].tolist()
        plt.text(x, y, str(rank[i]), color=f"C{n}", fontsize=7, fontweight="bold", ha="center", va="center")
plt.axis("off")
plt.show()
```

```python
# The first 12 characters in reading order, cropped with their own boxes
fig, axes = plt.subplots(1, 12, figsize=(12, 1.6))
for ax, i in zip(axes, reading_order[:12]):
    x1, y1, x2, y2 = r.boxes.xyxy[i].int().tolist()
    ax.imshow(r.orig_img[y1:y2, x1:x2, ::-1])
    ax.axis("off")
plt.show()
```
